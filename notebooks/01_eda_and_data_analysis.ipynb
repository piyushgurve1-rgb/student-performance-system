# Phase 1: Exploratory Data Analysis (EDA) & Dataset Exploration
### Project: Smart Student Dropout Prediction & Performance Analysis
**Dataset:** UCI Machine Learning Repository - Student Dropout and Academic Success (ID: 697)

This notebook documents the manual data exploration, cleaning, statistical profiling, and feature analysis performed during Phase 1.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual aesthetics
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
sns.set_palette('crest')
print('All libraries imported successfully!')
```

## 1. Data Ingestion & Shape Verification

```python
DATA_PATH = os.path.join('..', 'data', 'raw', 'data.csv')
df = pd.read_csv(DATA_PATH, sep=';')
print(f'Total Records (Rows): {df.shape[0]}')
print(f'Total Features (Columns): {df.shape[1]}')
df.head()
```

## 2. Missing Value & Data Type Audit

```python
missing_count = df.isnull().sum().sum()
print(f'Total Missing/Null Values across all cells: {missing_count}')
df.dtypes.value_counts()
```

## 3. Target Distribution & Class Imbalance Analysis

```python
target_counts = df['Target'].value_counts()
target_pct = df['Target'].value_counts(normalize=True) * 100

fig, ax = plt.subplots(figsize=(8, 4.5))
bars = ax.bar(target_counts.index, target_counts.values, color=['#10B981', '#EF4444', '#F59E0B'])
ax.set_title('Target Distribution: Class Imbalance Audit', fontsize=14, fontweight='bold', pad=15)
ax.set_ylabel('Number of Students', fontsize=12)
for bar, pct in zip(bars, target_pct):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 30, f'{int(height)} ({pct:.1f}%)',
            ha='center', va='bottom', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Key Socio-Economic Risk Factors: Tuition Fees & Scholarship Status

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Tuition fees up to date vs Target
sns.countplot(x='Tuition fees up to date', hue='Target', data=df, palette={'Dropout': '#EF4444', 'Graduate': '#10B981', 'Enrolled': '#F59E0B'}, ax=ax1)
ax1.set_title('Tuition Fees Paid (1) vs Defaulted (0)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Tuition Fees Up To Date (0 = In Debt, 1 = Cleared)')

# Scholarship Holder vs Target
sns.countplot(x='Scholarship holder', hue='Target', data=df, palette={'Dropout': '#EF4444', 'Graduate': '#10B981', 'Enrolled': '#F59E0B'}, ax=ax2)
ax2.set_title('Scholarship Holder (1) vs Non-Scholarship (0)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Scholarship Status (0 = No, 1 = Yes)')

plt.tight_layout()
plt.show()
```

## 5. Academic Performance (1st Semester Grade Distribution)

```python
plt.figure(figsize=(9, 4.5))
sns.kdeplot(data=df, x='Curricular units 1st sem (grade)', hue='Target', common_norm=False, fill=True, alpha=0.3,
            palette={'Dropout': '#EF4444', 'Graduate': '#10B981', 'Enrolled': '#F59E0B'})
plt.title('1st Semester Grade Density by Outcome Category', fontsize=13, fontweight='bold')
plt.xlabel('Average Grade in 1st Semester (Scale: 0 to 20)')
plt.tight_layout()
plt.show()
```
